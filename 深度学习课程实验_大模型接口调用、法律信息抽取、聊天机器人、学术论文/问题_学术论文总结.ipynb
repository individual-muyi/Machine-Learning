**定义一个路径整理函数，用来将论文名称和路径存入一个列表中，为之后解析文件做准备**

```python
def list_files_with_paths(directory):
    try:
        with os.scandir(directory) as entries:
            # entry.is_file() 检查当前条目是否为文件
            # os.path.splitext(entry.name)[0] 获取文件名（索引 0 代表只取文件名，不取扩展名）
            # entry.path 获取当前条目的完整路径
            files = [(os.path.splitext(entry.name)[0], entry.path) for entry in entries if entry.is_file()]
        return files
        
    # 捕获 FileNotFoundError 异常，当指定的目录不存在时会触发该异常
    except FileNotFoundError:
        # 打印错误信息，提示指定的目录不存在
        print(f"The directory '{directory}' does not exist.")
        # 返回一个空列表，表示没有找到任何文件
        return []

files = list_files_with_paths(directory_path)

# 检查一下操作是否正确
for file in files:
    print(file)
```

## 任务 3：通过批量（Batch）方法总结多个论文

**通过任务耗时能看出，论文总结花费的时间长，资源消耗多。如果我们要一次性总结大量论文，需要等待很长时间**

**现在介绍一种批量（Batch API）模式，用于无需实时反馈并需要处理大量请求的场景。**
- 优点：价格更低；无并发限制
- 缺点：需要额外操作；任务结束时间无法准确预估
- 学习文档：https://www.bigmodel.cn/dev/howuse/batchapi

**调用Batch API 必须实名认证，请先前往`实名认证`页面完成个人认证**
- https://open.bigmodel.cn/usercenter/settings/auth

**首先，创建一个用来批量处理的 Batch 文件**

- 信息：每个请求占据一行** 
- 格式：.jsonl——包含多个 JSON 对象的文件格式。每个 JSON 对象占据一行，行之间没有分割符号** 

```python
batch_requests = []

# 回顾一下先前创建的内容列表 file_contents，每个元素包含文件名 filename 和内容 content
for file in file_contents:

    # 为每个文件创建批量处理请求
    request = {
        "custom_id": f"{file['filename'][:64]}",    # custom id 最长 64
        "method": "POST",
        "url": "/v4/chat/completions",
        "body": {
            "model": model,    #每个batch文件只能包含对单个模型的请求
            "messages": [
                {"role": "system", "content": "角色：论文专家，擅长解读论文"},
                {
                    "role": "user",
                    "content": get_prompt(file['content'])
                }
            ],
            "temperature": 0.1
        }
    }
    
    batch_requests.append(request)

print(f"当前 batch 大小: {len(batch_requests)}")
```

**写入 JSONL 文件**

```python
output_file = Path("batch_requests.jsonl")  # 输出的 JSONL 文件路径

with output_file.open('w', encoding='utf-8') as f:
    for request in batch_requests:
        f.write(json.dumps(request, ensure_ascii=False) + "\n")

print(f"批量请求 JSONL 文件已生成：{output_file}")
```

**使用代码调用batch接口，先上传 batch 待处理的文件，再创建 batch 任务，这将会得到一个 batch_id**

```python
# 上传 batch 文件
result = client.files.create(
    file=open("batch_requests.jsonl", "rb"),
    purpose="batch"
)
print(result.id)

# 创建 batch 任务
create = client.batches.create(
    input_file_id=result.id,
    endpoint="/v4/chat/completions", 
    auto_delete_input_file=True,
    metadata={
        "description": "学术论文总结报告"
    }
)

# 查看 batch 任务的信息
print(create)
```

**通过 batch_id 查看文件当前的进度，若`status`字段为 completed 则完成了批量处理，否则需要等待直到任务结束**

**注意：Batch 由开放平台在闲时进行处理，预计需要等待一天时间，请于第二天查看返回结果**
