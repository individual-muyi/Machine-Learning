**初步实现了流畅的多轮对话，但是 Markdown 格式的内容还是没有清晰地呈现，需要借助 display 方法**

- 由于 display 方法并不存在像 print 中 (end='')的参数，用 display(Markdown()) 直接展示输出结果，token 间会换行
- 所以需要把所有输出拼在一起后再 display，进而需要一个变量 conversation_history 来实时记录对话历史
- **conversation_history 和 message 的区分：** conversation_history 用来记录流式输出过程中每一个时点的历史信息，用来呈现信息；message 只记录完整回答，用来向模型请求回答

````python
def Chatbotpro():
    # 创建 ZhipuAI 客户端实例，需要传入 API 密钥和基础 URL
    client = ZhipuAI(
        api_key=api_key,
        base_url=base_url
    )

    # 初始化消息列表，添加系统提示信息，此系统提示告知模型以热情友好且精准的方式解答用户问题
    message = [
        {
            "role": "system",
            "content": "你是一个乐于解答各种问题的助手，你的任务是为用户提供专业、准确、有见地的建议。"
        }
    ]

    # 初始化对话历史记录字符串，用于存储整个对话过程
    conversation_history = ""

    while True:
        # 定义用户输入分隔字符串，用于在输出中清晰区分用户输入部分
        string_user = """========================================**用户输入**======================================="""
        # 以 Markdown 格式显示用户输入分隔字符串
        display(Markdown(string_user))

        # 获取用户输入
        Prompt = input()

        # 如果用户输入 'exit'，则退出 Chatbotpro 函数
        if Prompt == 'exit':
            return

        # 将用户输入信息添加到对话历史记录中
        # 包含用户输入分隔字符串和实际输入内容
        conversation_history += f"{string_user}\n\n{Prompt}\n"

        # 将用户输入信息添加到消息列表中，以字典形式表示
        # "role" 为 "user" 表示这是用户的消息，"content" 存储用户输入内容
        message.append({"role": "user", "content": Prompt})

        # 定义模型输出分隔字符串，用于在输出中清晰区分模型输出部分
        string_model = """========================================**模型输出**======================================"""
        
        # 将模型输出分隔字符串添加到对话历史记录中
        conversation_history += f"\n{string_model}\n\n"

        # 以 Markdown 格式显示更新后的对话历史记录，并清空之前的显示内容
        display(Markdown(conversation_history), clear=True)

        #### 调用模型进行对话，开启流式输出
        completion = ____________________

        # 初始化完整回复字符串，用于存储模型的完整回复
        full_response = ""

        # 遍历模型返回的每一个数据块
        for chunk in completion:
            # 提取当前数据块中的回复内容
            content = chunk.choices[0].delta.content
            if content:
                # 将当前数据块的内容添加到完整回复字符串中
                full_response += content
                
                #### 以 Markdown 格式显示更新后的对话历史记录和当前完整回复，并清空之前的显示内容，实现流式输出效果
                display(Markdown(), clear=____________________)

        #### 将模型的完整回复添加到对话历史记录中
        conversation_history += ____________________

        # 将模型的完整回复添加到消息列表中，以字典形式表示
        # "role" 为 "assistant" 表示这是模型的回复，"content" 存储模型的完整回复内容
        message.append({"role": "assistant", "content": full_response})


#### 调用 Chatbotpro 函数开始对话，给聊天机器人设定一个角色回答问题
Chatbotpro()
````

**将新的消息添加到对话历史记录中**

- history 是一个列表，列表中的每个元素是一个元组，元组的第一个元素是用户输入的消息文本，第二个元素是模型回复的消息文本。例如：
```
history = [
    ("你好\n", "你好呀！有什么我可以帮忙的吗？"),
    ("介绍一下大模型\n", None)
]
```
- **记录用户输入：** 当用户输入新消息时，add_message 函数会将用户输入的消息文本加上换行符，作为元组的第一个元素，元组的第二个元素初始化为 None，表示还没有模型的回复。然后将这个元组添加到 history 列表中。
- **更新界面状态：** 函数返回更新后的 history 列表和一个 Gradio 文本框对象，用于更新界面显示

````python
def add_message(history, message):
    """
    该函数用于将新的消息添加到对话历史记录中。

    参数:
    history (list): 存储对话历史的列表，列表中的每个元素是一个元组，
                    元组的第一个元素是用户或模型的消息文本，第二个元素可以是其他信息（这里暂时为 None）。
    message (dict): 包含新消息信息的字典，该字典至少应包含 "text" 键，用于存储消息文本。

    返回:
    tuple: 返回一个元组，元组的第一个元素是更新后的对话历史记录列表，
           第二个元素是一个 Gradio 文本框对象，文本框内无内容且不支持上传附件（interactive=False）。
    """
    # 检查传入的消息字典中的 "text" 键对应的值是否不为 None
    if message["text"] is not None:
        # 如果消息文本不为 None，则将消息文本加上换行符后作为元组的第一个元素，
        # 元组的第二个元素设置为 None，然后将该元组添加到对话历史记录列表中
        history.append((message["text"] + "\n", None))

    # 返回更新后的对话历史记录列表和一个文本框，文本框内无内容且不支持上传附件（interactive=False）
    return history, gr.MultimodalTextbox(value=None, interactive=False)
````

````python
# 该函数用于处理多轮对话
def bot(history):
    # 将当前用户输入对应的回复初始化为空字符串，准备接收模型的回复
    history[-1][1] = ""

    # 将历史对话转换为 ZhipuAI 格式的消息列表
    history_zhipuai_format = []
    # 遍历历史对话中的每一轮（除了当前轮）
    for human, assistant in history[:-1]:
        # 添加用户消息到消息列表
        history_zhipuai_format.append({"role": "user", "content": human})
        # 添加助手消息到消息列表
        history_zhipuai_format.append({"role": "assistant", "content": assistant})
    
    #### 补全代码，添加当前用户的消息到消息列表
    history_zhipuai_format.append({____________________})

    # 调用 Chat 函数进行对话，获取流式响应
    response = Chat(history_zhipuai_format)

    # 遍历流式响应的每个数据块
    for chunk in response:
        # 将当前数据块的内容添加到当前回复中
        history[-1][1] += chunk.choices[0].delta.content
        # 使用生成器逐块返回更新后的历史对话，实现流式输出
        yield history
````

**创建 Gradio 界面**

````python
with gr.Blocks() as demo:
    # 创建聊天机器人组件，用于显示聊天历史
    chatbot = gr.Chatbot([], elem_id="聊天历史", bubble_full_width=False, label="聊天历史")

    # 创建多模态文本输入框，用于用户输入消息，不支持上传附件（interactive=False）
    chat_input = gr.MultimodalTextbox(
        interactive=False,
        placeholder="Enter message...",
        show_label=False,
        elem_id="chat_input"
    )

    # 当用户提交消息时，调用 add_message 函数将消息添加到聊天历史中
    chat_msg = chat_input.submit(add_message, [chatbot, chat_input], [chatbot, chat_input])

    # 当聊天消息处理完毕后，调用 bot 函数进行多轮对话处理
    bot_msg = chat_msg.then(bot, [chatbot], chatbot, api_name="bot_response")

    # 创建一个按钮来清空聊天历史
    clear_button = gr.Button("清空聊天历史")
    # 当按钮被点击时，调用 clear 函数将聊天历史设置为空列表
    clear_button.click(clear, None, chatbot)

    # 对话处理完成后，方便用户继续输入
    bot_msg.then(lambda: gr.MultimodalTextbox(interactive=False), None, [chat_input])

    # 提供问题样例，方便用户测试
    gr.Examples(examples=[{"text": "你是谁？"}, {"text": "介绍一下大模型"}, {"text": "AI领域的提示词prompt是什么？"}, {"text": "我总共问了哪些问题？"}], inputs=chat_input, elem_id="chat_input")


if __name__ == "__main__":
    # 启动 Gradio 应用，设置默认并发限制为 10，不开启共享，显示错误信息
    demo.queue(default_concurrency_limit=10).launch(share=False, show_error=True)
````
